```python
import pandas as pd
df = pd.read_csv('/student-mat.csv')
print(df.head())
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df.shape
```

```python
df_cleaned = df.dropna()
df_filled = df_cleaned.fillna(df.median(numeric_only=True))
print(df_filled)
```

```python
average_g3 = df_filled["G3"].mean()
print("Average G3 score:", average_g3)
```

```python
high_achievers = df_filled[df_filled["G3"] > 15]
count_high_achievers = high_achievers.shape[0]

print("Number of students scoring above 15 in G3:", count_high_achievers)
```

```python
correlation = df_filled["studytime"].corr(df_filled["G3"])
print("Correlation between study time and G3:", correlation)
```

```python
avg_g3_by_gender = df_cleaned.groupby("sex")["G3"].mean()
print(avg_g3_by_gender)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df_cleaned["G3"], bins=10, edgecolor="black")
plt.title("Distribution of Final Grades (G3)")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.scatter(df_cleaned["studytime"], df_cleaned["G3"], alpha=0.6)
plt.title("Study Time vs Final Grade (G3)")
plt.xlabel("Study Time (1=low, 4=high)")
plt.ylabel("Final Grade (G3)")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

```python
import matplotlib.pyplot as plt


avg_g3_by_gender = df_cleaned.groupby("sex")["G3"].mean()
plt.figure(figsize=(6,5))
avg_g3_by_gender.plot(kind="bar", color=["skyblue", "lightcoral"], edgecolor="black")

plt.title("Average Final Grade (G3) by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade (G3)")
plt.xticks(rotation=0)
plt.show()
```
